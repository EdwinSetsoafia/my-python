```python
name = input("Enter your full name: ")
age = int(input("Enter your age: "))

# Check if student is an adult or minor
if age < 18:
    age_status = "Minor"
else:
    age_status = "Adult"

gpa = float(input("Enter your GPA: "))

# Check Dean's List status
if gpa > 4.0:
    print("Invalid GPA")
    deans_list = "No"
elif gpa >= 3.5:
    deans_list = "Yes"
else:
    deans_list = "No"

num_courses = int(input("How many courses are you taking? "))

courses_list = []

for i in range(num_courses):
    course = input(f"Enter course {i + 1}: ")
    courses_list.append(course)


student_profile = {
    "name": name,
    "age": age,
    "gpa": gpa,
    "courses": courses_list,
    "num_courses": num_courses
}

print("\n===== STUDENT PROFILE =====")
print(f"Name: {student_profile['name']}")
print(f"Age: {student_profile['age']} ({age_status})")
print(f"GPA: {student_profile['gpa']}")
print(f"Courses: {student_profile['courses']}")
print(f"Number of Courses: {student_profile['num_courses']}")
print(f"Dean's List: {deans_list}")
print("===========================")

```
